```python
from collections import deque

# --- Helper functions for adjacency matrix ---
def createMatrix(n):
    m = []
    for i in range(n):
        c = []
        for j in range(n):
            c.append(0)
        m.append(c)
    return m

def printMatrix(m):
    for r in m:
        print(r)
    print()  # extra newline

# --- Read graph from file ---
f = open("g1.txt", "r")
vertex_edges = f.readline()
vertex_edges_list = vertex_edges.split()
number_of_vertex = int(vertex_edges_list[0])
number_of_edge = int(vertex_edges_list[1])

# Create adjacency matrix
G = createMatrix(number_of_vertex)

# Read edges and fill adjacency matrix
for _ in range(number_of_edge):
    u, v = map(int, f.readline().split())
    G[u][v] = 1
    G[v][u] = 1  # undirected graph

f.close()

# --- Print adjacency matrix ---
print("Adjacency Matrix:")
printMatrix(G)

# --- DFS traversal using recursion ---
visited_dfs = [False] * number_of_vertex
def dfs_recursive(G, start):
    visited_dfs[start] = True
    print(start, end=" ")
    for neighbor in range(number_of_vertex):
        if G[start][neighbor] == 1 and not visited_dfs[neighbor]:
            dfs_recursive(G, neighbor)

print("DFS Traversal:")
dfs_recursive(G, 0)  # start from node 0
print("\n")

# --- BFS traversal using queue ---
visited_bfs = [False] * number_of_vertex
def bfs(G, start):
    q = deque()
    q.append(start)
    visited_bfs[start] = True

    while q:
        node = q.popleft()
        print(node, end=" ")
        for neighbor in range(number_of_vertex):
            if G[node][neighbor] == 1 and not visited_bfs[neighbor]:
                visited_bfs[neighbor] = True
                q.append(neighbor)

print("BFS Traversal:")
bfs(G, 0)
print()
```

Task 6 — Count Connected Components
Modify DFS to count how many connected components the graph has.

```python
# Number of nodes and edges
n, e = map(int, input().split())

# Graph as adjacency list
g = [[] for _ in range(n + 1)]
visited = [False] * (n + 1)

# Read edges
for _ in range(e):
    u, v = map(int, input().split())
    g[u].append(v)
    g[v].append(u)

# DFS function
def dfs(source):
    visited[source] = True
    for child in g[source]:
        if not visited[child]:
            dfs(child)

# Count connected components
cnt = 0
for i in range(1, n + 1):
    if not visited[i]:
        dfs(i)
        cnt += 1

print(f"Number of connected components: {cnt}")
```

Task 7 — Detect Cycle in an Undirected Graph
Write a DFS-based algorithm that returns True/False if a cycle exists.

```python
# Number of nodes and edges
n, e = map(int, input().split())

# Graph as adjacency list
g = [[] for _ in range(n + 1)]
visited = [False] * (n + 1)

# Read edges
for _ in range(e):
    u, v = map(int, input().split())
    g[u].append(v)
    g[v].append(u)

# DFS function to detect cycle
def dfs(node, parent):
    visited[node] = True
    for child in g[node]:
        if not visited[child]:
            if dfs(child, node):  # recursive call
                return True
        else:
            if child != parent:  # visited node is not parent → cycle
                return True
    return False

# Call DFS starting from node 1
if dfs(1, -1):
    print("YES")
else:
    print("NO")
```
