Task 2 — Write DFS Using Recursion
Rewrite the DFS function in a recursive format:
def dfs_recursive(G, start):

```python
def dfs_recursive(G, start, visited=None):
    if visited is None:
        visited = set()

    print(start, end=" ")
    visited.add(start)

    for v in G[start]:
        if v not in visited:
            dfs_recursive(G, v, visited)
```

Task 3 — Modify DFS to Track Path
Modify DFS so it prints the path from the start node to each visited node.
Example Output:
Path to 3: 0 → 1 → 3
Path to 5: 0 → 2 → 4 → 5

```python
def dfs_paths(G, start, path=None, visited=None):
    if visited is None:
        visited = set()
    if path is None:
        path = [start]

    visited.add(start)

    print("Path to", start, ":", " → ".join(map(str, path)))

    for v in G[start]:
        if v not in visited:
            dfs_paths(G, v, path + [v], visited)
```

Task 4 — Implement BFS Using a Queue
Write a BFS algorithm using Python queue.Queue() for both:
adjacency matrix
adjacency list
Expected function signature:
def bfs(G, start):

```python
import queue

def bfs_matrix(G, start):
    q = queue.Queue()
    visited = set()

    q.put(start)
    visited.add(start)

    while not q.empty():
        u = q.get()
        print(u, end=" ")

        for v in range(len(G)):
            if G[u][v] == 1 and v not in visited:
                visited.add(v)
                q.put(v)
```

Adjacency List Version

```python
import queue

def bfs_list(G, start):
    q = queue.Queue()
    visited = set()

    q.put(start)
    visited.add(start)

    while not q.empty():
        u = q.get()
        print(u, end=" ")

        for v in G[u]:
            if v not in visited:
                visited.add(v)
                q.put(v)
```

Task 5 — Compare DFS and BFS Outputs
For a given graph in g.txt, print both DFS and BFS traversal:
Which one goes deeper first?
Which one finds the shortest path in an unweighted graph?

```python
def load_graph(filename):
    f = open(filename)
    n, e = map(int, f.readline().split())

    # adjacency matrix
    mat = [[0]*n for _ in range(n)]

    # adjacency list
    lst = [[] for _ in range(n)]

    for line in f:
        u, v = map(int, line.split())

        mat[u][v] = 1
        mat[v][u] = 1   # undirected

        lst[u].append(v)
        lst[v].append(u)

    f.close()
    return mat, lst
```
