🧠 1. List Operations

```python
numbers = [10, 20, 20, 30, 40, 10, 50]
#Here we create a list of unique numbers

unique_numbers = []
for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)


max_num = unique_numbers[0]
min_num = unique_numbers[0]
total = 0

for n in unique_numbers:
    if n > max_num:
        max_num = n
    if n < min_num:
        min_num = n
    total += n

avg = total / len(unique_numbers)

print("Unique:", unique_numbers)
print("Max:", max_num)
print("Min:", min_num)
print("Average:", avg)
```

🧠 13. List Processing with Function

```python
def square_list(nums):
    squares = []
    for n in nums:
        square = n * n
        squares.append(square)
    return squares

numbers = [1, 2, 3, 4, 5]
result = square_list(numbers)
print(result)
```

🧠 20. OOP – Stack Class

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if not self.stack:
            return "Stack is empty"
        return self.stack.pop()

    def display(self):
        print("Stack:", self.stack)


# Demonstration
s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.display()
s.pop()
s.display()
```

```python
# Task 2: Sets and Dictionaries

# Create a set with duplicates
my_set = {1, 2, 2, 3, 4, 4, 5}
print("Set with duplicates removed automatically:", my_set)

# Create dictionary
student_marks = {"Alice": 85, "Bob": 78, "Charlie": 92}

# Add one more student
student_marks["David"] = 88

# Print all keys
print("\nAll keys:", list(student_marks.keys()))

# Print all values
print("All values:", list(student_marks.values()))

# Find the student with highest marks
highest_student = max(student_marks, key=student_marks.get)
print("Highest marks obtained by:", highest_student, "=", student_marks[highest_student])
```

Task 3: String Frequency Counter

```python
# Task 3: String Frequency Counter

text = input("Enter a string: ")

count = {}

for char in text:
    if char.isalpha():        # count only alphabetic characters
        char = char.lower()   # ignore case (optional)
        count[char] = count.get(char, 0) + 1

# Sort the dictionary by frequency (highest first)
sorted_count = dict(sorted(count.items(), key=lambda x: x[1], reverse=True))

print("\nCharacter frequency (sorted):")
for char, freq in sorted_count.items():
    print(f"{char}: {freq}")
```
