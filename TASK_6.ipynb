Task 3 — Modify DFS to Track Path
Modify DFS so it prints the path from the start node to each visited node.
Example Output:
Path to 3: 0 → 1 → 3
Path to 5: 0 → 2 → 4 → 5

```python
# Open file
f = open("g.txt", "r")

# Read number of vertices and edges
vertex_edges = f.readline()
vertex_edges_list = vertex_edges.split()
number_of_vertex = int(vertex_edges_list[0])
number_of_edge = int(vertex_edges_list[1])

# Create graph and visited list
G = [[]] * (number_of_vertex + 1)
visited = [False] * (number_of_vertex + 1)
parent = [-1] * (number_of_vertex + 1)

# Read edges
for _ in range(number_of_edge):
    edge = f.readline()
    edge_list = edge.split()
    u = int(edge_list[0])
    v = int(edge_list[1])
    G[u].append(v)
    G[v].append(u)

# DFS with parent tracking
def dfs_recursive(G, start):
    visited[start] = True
    for neighbor in G[start]:
        if not visited[neighbor]:
            parent[neighbor] = start
            dfs_recursive(G, neighbor)

# Start DFS from node 1
dfs_recursive(G, 0)

# Function to print path
def print_path(node):
    path = []
    while node != -1:
        path.append(node)
        node = parent[node]
    path.reverse()
    return path

# Print path to each visited node
for i in range(0, number_of_vertex + 1):
    if visited[i]:
        p = print_path(i)
        print("Path to", i, ":", end=" ")
        for j in range(len(p)):
            if j > 0:
                print("→", end=" ")
            print(p[j], end=" ")
        print()

# Close file
f.close()
```

Task 6 — Count Connected Components
Modify DFS to count how many connected components the graph has.

```python
# Number of nodes and edges
n, e = map(int, input().split())

# Graph as adjacency list
g = [[] for _ in range(n + 1)]
visited = [False] * (n + 1)

# Read edges
for _ in range(e):
    u, v = map(int, input().split())
    g[u].append(v)
    g[v].append(u)

# DFS function
def dfs(source):
    visited[source] = True
    for child in g[source]:
        if not visited[child]:
            dfs(child)

# Count connected components
cnt = 0
for i in range(1, n + 1):
    if not visited[i]:
        dfs(i)
        cnt += 1

print(f"Number of connected components: {cnt}")
```

Task 7 — Detect Cycle in an Undirected Graph
Write a DFS-based algorithm that returns True/False if a cycle exists.

```python
# Number of nodes and edges
n, e = map(int, input().split())

# Graph as adjacency list
g = [[] for _ in range(n + 1)]
visited = [False] * (n + 1)

# Read edges
for _ in range(e):
    u, v = map(int, input().split())
    g[u].append(v)
    g[v].append(u)

# DFS function to detect cycle
def dfs(node, parent):
    visited[node] = True
    for child in g[node]:
        if not visited[child]:
            if dfs(child, node):  # recursive call
                return True
        else:
            if child != parent:  # visited node is not parent → cycle
                return True
    return False

# Call DFS starting from node 1
if dfs(1, -1):
    print("YES")
else:
    print("NO")
```
